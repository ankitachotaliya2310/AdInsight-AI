# Section 2 – Marketing Channel Analysis

Which marketing channels generate the highest number of website sessions?

Understanding traffic acquisition channels helps businesses identify where users are coming from and where marketing efforts should be focused.

Examples of marketing channels include:

- Organic Search
- Direct
- Referral
- Social
- Paid Search

```python
# =====================================================
# Traffic by Marketing Channel
# =====================================================

traffic_channel = (
    df.groupby("channelGrouping")
      .size()
      .reset_index(name="Sessions")
      .sort_values("Sessions", ascending=False)
)

# Calculate percentage contribution
traffic_channel["Percentage"] = (
    traffic_channel["Sessions"] /
    traffic_channel["Sessions"].sum()
) * 100

traffic_channel["Percentage"] = traffic_channel["Percentage"].round(2)

traffic_channel
```

# Section 3 – Traffic Source Analysis

## Business Question

Which traffic sources generate the highest number of website sessions?

### Why is this important?

Traffic source analysis helps businesses understand where visitors are coming from.

Instead of looking at broad marketing channels, this analysis identifies the exact websites or search engines responsible for bringing users to the platform.

Examples include:

- Google
- YouTube
- Facebook
- Bing
- Yahoo
- Direct traffic

```python
# =====================================================
# Top Traffic Sources
# =====================================================

traffic_source = (
    df.groupby("source")
      .size()
      .reset_index(name="Sessions")
      .sort_values("Sessions", ascending=False)
      .head(10)
)

traffic_source["Percentage"] = (
    traffic_source["Sessions"] /
    traffic_source["Sessions"].sum()
    * 100
).round(2)

traffic_source
```

```python
# =====================================================
# Top 10 Website Traffic Sources
# Executive Dashboard Style
# =====================================================

import plotly.graph_objects as go

# -----------------------------------------------------
# Prepare Data
# -----------------------------------------------------

traffic_source = (
    df.groupby("source")
      .size()
      .reset_index(name="Sessions")
      .sort_values("Sessions", ascending=False)
      .head(10)
      .reset_index(drop=True)
)

# Calculate Traffic Share (%)
traffic_source["Share"] = (
    traffic_source["Sessions"] /
    traffic_source["Sessions"].sum() * 100
).round(1)

# Rank
traffic_source["Rank"] = (
    traffic_source.index + 1
)

# Labels
traffic_source["Label"] = (
    traffic_source["Sessions"]
    .map("{:,.0f}".format)
    + " ("
    + traffic_source["Share"].astype(str)
    + "%)"
)

# -----------------------------------------------------
# Google Material Color Palette
# -----------------------------------------------------

colors = [
    "#1967D2",  # Google Blue (Top Source)
    "#669DF6",
    "#8AB4F8",
    "#D2E3FC",
    "#DADCE0",
    "#DADCE0",
    "#DADCE0",
    "#DADCE0",
    "#DADCE0",
    "#DADCE0"
]

# -----------------------------------------------------
# Create Figure
# -----------------------------------------------------

fig = go.Figure()

fig.add_trace(

    go.Bar(

        x=traffic_source["Sessions"],

        y=traffic_source["source"],

        orientation="h",

        marker=dict(

            color=colors,

            line=dict(
                color="white",
                width=1
            )

        ),

        text=traffic_source["Label"],

        textposition="auto",

        textfont=dict(
            family="Arial",
            size=13
        ),

        cliponaxis=False,

        customdata=traffic_source[
            ["Share", "Rank"]
        ],

        hovertemplate=
        "<b>Traffic Source</b><br>"
        "%{y}<br><br>"
        "<b>Sessions</b><br>"
        "%{x:,}<br><br>"
        "<b>Traffic Share</b><br>"
        "%{customdata[0]}%<br><br>"
        "<b>Rank</b><br>"
        "#%{customdata[1]}"
        "<extra></extra>"
    )

)

# -----------------------------------------------------
# Layout
# -----------------------------------------------------

fig.update_layout(

    template="simple_white",

    title=dict(

        text="<b>Top 10 Website Traffic Sources</b>"
             "<br><sup>Google Analytics Sample Dataset • Top 10 Sources by Sessions</sup>",

        x=0.5,

        font=dict(
            size=28,
            family="Arial"
        )

    ),

    font=dict(

        family="Arial",

        size=16,

        color="#202124"

    ),

    plot_bgcolor="white",

    paper_bgcolor="white",

    xaxis=dict(

        title="",

        showgrid=True,

        gridcolor="#F1F3F4",

        zeroline=False,

        tickformat=",",

        range=[0, 40000]

    ),

    yaxis=dict(

        title=""

    ),

    margin=dict(

        l=130,

        r=60,

        t=130,

        b=60

    ),

    height=650,

    showlegend=False,

    hoverlabel=dict(

        bgcolor="white",

        bordercolor="#DADCE0",

        font_size=13,

        font_family="Arial"

    )

)
fig.update_traces(
    width=0.75
)
# -----------------------------------------------------
# Display Chart
# -----------------------------------------------------

fig.show()
```

# 🌍 Section 4 – Geographic Analysis

## Business Question

Which countries generate the highest website traffic and revenue?

### Why is this important?

Geographic analysis helps businesses understand where their customers are located. These insights support regional marketing strategies, localization, international expansion, and advertising budget allocation.

This section analyzes:

- Top countries by website sessions
- Top countries by unique users
- Revenue generated by each country
- Geographic distribution of website traffic

```python
# =====================================================
# Geographic Performance
# =====================================================

country_summary = (
    df.groupby("country")
      .agg(
          Sessions=("fullVisitorId", "count"),
          Users=("fullVisitorId", "nunique"),
          Revenue=("transactionRevenue",
                   lambda x: x.fillna(0).sum()/1_000_000)
      )
      .reset_index()
)

country_summary["Revenue"] = country_summary["Revenue"].round(2)

country_summary = (
    country_summary
    .sort_values("Sessions", ascending=False)
)

country_summary.head(10)
```

```python
top10_country = country_summary.head(10)

top10_country
```

```python
# =====================================================
# Top 10 Countries by Website Sessions
# Executive Dashboard Style
# =====================================================

import plotly.graph_objects as go

# -----------------------------------------------------
# Prepare Data
# -----------------------------------------------------

top10_country = (
    country_summary
    .sort_values("Sessions", ascending=False)
    .head(10)
    .reset_index(drop=True)
)

top10_country["Share"] = (
    top10_country["Sessions"] /
    top10_country["Sessions"].sum() * 100
).round(1)

top10_country["Rank"] = top10_country.index + 1

top10_country["Label"] = (
    top10_country["Sessions"]
        .map("{:,.0f}".format)
    + " ("
    + top10_country["Share"].astype(str)
    + "%)"
)

# -----------------------------------------------------
# Google Material Colors
# -----------------------------------------------------

colors = [
    "#1967D2",
    "#8AB4F8",
    "#8AB4F8",
    "#D2E3FC",
    "#DADCE0",
    "#DADCE0",
    "#DADCE0",
    "#DADCE0",
    "#DADCE0",
    "#DADCE0"
]

# -----------------------------------------------------
# Figure
# -----------------------------------------------------

fig = go.Figure()

fig.add_trace(

    go.Bar(

        x=top10_country["Sessions"],

        y=top10_country["country"],

        orientation="h",

        marker=dict(
            color=colors,
            line=dict(color="white", width=1)
        ),

        text=top10_country["Label"],

        textposition="auto",

        textfont=dict(
            family="Arial",
            size=13
        ),

        width=0.72,

        cliponaxis=False,

        customdata=top10_country[
            ["Users","Revenue","Share","Rank"]
        ],

        hovertemplate=
        "<b>%{y}</b><br><br>"
        "Sessions : %{x:,}<br>"
        "Users : %{customdata[0]:,}<br>"
        "Revenue : $%{customdata[1]:,.2f}M<br>"
        "Traffic Share : %{customdata[2]}%<br>"
        "Rank : #%{customdata[3]}"
        "<extra></extra>"

    )

)

fig.update_layout(

    template="simple_white",

    title=dict(
        text="<b>Top 10 Markets by Website Sessions</b><br>"
             "<sup>Google Analytics Sample Dataset • Geographic Traffic Distribution</sup>",
        x=0.5,
        font=dict(size=28)
    ),

    height=650,

    margin=dict(
        l=150,
        r=60,
        t=120,
        b=40
    ),

    showlegend=False,

    font=dict(
        family="Arial",
        size=14,
        color="#202124"
    ),

    plot_bgcolor="white",
    paper_bgcolor="white",

    hoverlabel=dict(
        bgcolor="white",
        bordercolor="white",
        font_size=13
    )

)

fig.update_xaxes(

    title="",

    tickformat=",.0f",

    gridcolor="#F1F3F4",

    zeroline=False

)

fig.update_yaxes(

    title="",

    categoryorder="total ascending"

)

fig.show()
```

```python
# =====================================================
# Campaign Performance Summary
# =====================================================

campaign_summary = (

    df.groupby("campaign")

      .agg(

          Sessions=("fullVisitorId","count"),

          Users=("fullVisitorId","nunique"),

          Transactions=("transactions",
                        lambda x: x.fillna(0).sum()),

          Revenue=("transactionRevenue",
                   lambda x: x.fillna(0).sum()/1_000_000)

      )

      .reset_index()

)

campaign_summary["Revenue"] = campaign_summary["Revenue"].round(2)

campaign_summary = campaign_summary.sort_values(
    "Sessions",
    ascending=False
)

campaign_summary
```
